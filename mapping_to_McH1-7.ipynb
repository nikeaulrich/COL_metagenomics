## Mapping reads against McH1-7 genome

https://www.ncbi.nlm.nih.gov/datasets/genome/?bioproject=PRJNA639770

```python
#!/bin/bash
#SBATCH -c 24  # Number of Cores per Task
#SBATCH --mem=100G  # Requested Memory
#SBATCH -p cpu  # Partition
#SBATCH -t 24:00:00  # Job time limit
#SBATCH --mail-type=ALL
#SBATCH -o /work/pi_sarah_gignouxwolfsohn_uml_edu/nikea/COL/mch17_blast/slurm-mch17-mapping-%j.out  # %j = job ID

module load conda/latest
conda activate anvio-8

SAMPLENAME="ofav"
READSPATH="/work/pi_sarah_gignouxwolfsohn_uml_edu/nikea/COL/assembly/${SAMPLENAME}"
CONTIGPATH='/work/pi_sarah_gignouxwolfsohn_uml_edu/nikea/COL/mch17_blast'
CONTIGFILE='GCF_013366255.1_ASM1336625v1_genomic_mch17_assembly.fna'

cd $CONTIGPATH
bowtie2-build $CONTIGFILE mch1_7_contigs

bowtie2 --threads 11 -x mch1_7_contigs -1 $READSPATH/ofav_reads_R1_ALL.fastq.gz -2 $READSPATH/ofav_reads_R2_ALL.fastq.gz -S mch1_7.sam

samtools view -F 4 -b -S mch1_7.sam -o mch1_7-RAW.bam

anvi-init-bam mch1_7-RAW.bam -o mch1_7.bam

rm mch1_7-RAW.bam

#JOB ID: 29222923 (ofav) changed and reran for mcav 29223414
#bash script: nikea/COL/bash_scripts/Col_mch17_mapping.sh
```
